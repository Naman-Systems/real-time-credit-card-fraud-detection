## Precision-Recall AUC & Decision Threshold Analysis

We evaluate candidates on validation and out-of-time test partitions, optimizing the decision threshold to maximize F1-score.


```python
from sklearn.metrics import classification_report, average_precision_score, roc_auc_score, confusion_matrix

X_val = val_df.drop(columns=['Class'])
y_val = val_df['Class'].values
X_val_trans = preprocessor.transform(X_val)

val_probs = rf.predict_proba(X_val_trans)[:, 1]
print(f'Validation PR-AUC: {average_precision_score(y_val, val_probs):.4f}')
print(f'Validation ROC-AUC: {roc_auc_score(y_val, val_probs):.4f}')

# Evaluate at optimal threshold (0.40)
y_pred_opt = (val_probs >= 0.40).astype(int)
print('\nClassification Report @ 0.40:')
print(classification_report(y_val, y_pred_opt))
print('\nConfusion Matrix:\n', confusion_matrix(y_val, y_pred_opt))
```
