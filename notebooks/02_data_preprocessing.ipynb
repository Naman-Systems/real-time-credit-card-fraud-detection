# 02: Data Preprocessing & Chronological Splitting
## Preventing Data Leakage in Fraud Detection Systems

This notebook demonstrates strict forward chronological partitioning (70% train, 15% validation, 15% test) and feature scaling using RobustScaler fitted solely on training data.


```python
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler

df = pd.read_csv('../dataset/creditcard.csv')
df_sorted = df.sort_values(by='Time').reset_index(drop=True)
n = len(df_sorted)
train_end = int(0.70 * n)
val_end = int(0.85 * n)

train_df = df_sorted.iloc[:train_end]
val_df = df_sorted.iloc[train_end:val_end]
test_df = df_sorted.iloc[val_end:]

print(f'Train size: {len(train_df)}, frauds: {train_df.Class.sum()}')
print(f'Val size: {len(val_df)}, frauds: {val_df.Class.sum()}')
print(f'Test size: {len(test_df)}, frauds: {test_df.Class.sum()}')
```

```python
X_train = train_df.drop(columns=['Class'])
y_train = train_df['Class']

pca_features = [f'V{i}' for i in range(1, 29)]
preprocessor = ColumnTransformer([
    ('amount_scaler', RobustScaler(), ['Amount']),
    ('time_scaler', RobustScaler(), ['Time']),
    ('pca_passthrough', 'passthrough', pca_features)
])

X_train_trans = preprocessor.fit_transform(X_train)
print('Transformed feature matrix shape:', X_train_trans.shape)
```
