# 01: Exploratory Data Analysis (EDA)
## Real-Time Credit Card Fraud Detection System

This notebook performs exploratory data analysis on the ULB Credit Card Fraud dataset.
We inspect shape, missing values, duplicates, class imbalance (0.172% fraud), amount distributions, and correlation patterns.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/creditcard.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
print('Missing values:', df.isnull().sum().sum())
print('Duplicates:', df.duplicated().sum())
class_counts = df['Class'].value_counts()
print(class_counts)
print(f'Fraud percentage: {class_counts[1] / len(df) * 100:.4f}%')
```

```python
# Amount statistics comparison
print('Legitimate transactions summary:')
print(df[df['Class'] == 0]['Amount'].describe().round(2))
print('\nFraudulent transactions summary:')
print(df[df['Class'] == 1]['Amount'].describe().round(2))
```

```python
# Visualizing Class Imbalance
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Class', palette=['#10B981', '#EF4444'])
plt.title('Transaction Class Counts (0: Legit, 1: Fraud)')
plt.yscale('log')
plt.show()
```
